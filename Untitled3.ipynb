```python
# Лабораторная работа 3
# ФИО:Соболева Диана Юрьевна
# Группа: ЦИБ-251
# Реализованные алгоритмы:
## Поиск: Линейный поиск (O(n)), Бинарный поиск (O(log n))
## Сортировка O(n²): Пузырьковая (Bubble Sort), Выбором (Selection Sort), Вставками (Insertion Sort)
## Улучшенная сортировка: Сортировка Шелла (Shell Sort) с шагом n//2
## Аналитика: Функции проверки отсортированности и подсчета обменов
def binary_search(arr, target):
    left = 0
    right = len(arr) - 1

    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1
```

```python
if __name__ == "__main__":
    result = binary_search([1, 3, 5, 7, 9, 11], 9)
    print(f"Результат поиска 9 в [1, 3, 5, 7, 9, 11]: {result}")
    print(binary_search([1, 3, 5, 7, 9, 11], 6))
    print(binary_search([], 5))
    print(binary_search([5], 5))
```

```python
def sort_sales_data(sales):
    n = len(sales)
    for i in range(n - 1):
        swapped = False
        for j in range(0, n - i - 1):
            if sales[j] < sales[j + 1]:
                sales[j], sales[j + 1] = sales[j + 1], sales[j]
                swapped = True
        if not swapped:
            break
    return sales
```
