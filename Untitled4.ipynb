```python
# Курс: ЦИБ - 251
# ФИО: Соболева Диана Юрьевна
# Задание 23: Реализовать функцию, которая сравнивает хеши двух подстрок одинаковой длины, взятых из разных позиций одной или разных строк, используя полиномиальный хэш. Это основа алгоритма Рабина-Карпа для быстрого сравнения без посимвольного перебора

def compare_substring_hashes(s1: str, start1: int, s2: str, start2: int, length: int, p: int = 31, mod: int = 10**9 + 7) -> bool:
    if start1 + length > len(s1) or start2 + length > len(s2):
        return False
    hash1 = 0
    hash2 = 0
    for i in range(length):
        hash1 = (hash1 * p + ord(s1[start1 + i])) % mod
    for i in range(length):
        hash2 = (hash2 * p + ord(s2[start2 + i])) % mod

    return hash1 == hash2
```
