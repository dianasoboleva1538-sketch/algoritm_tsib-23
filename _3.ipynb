```python
# Лабораторная работа 2
# Курс: ЦИБ - 251
# ФИО: Соболева Диана Юрьевна
## Описание варианта
### Граф
#- Вершины: 7 верш: (1,2), (1,3), (2,4), (3,5),(5,6), (6,7)
#- Рёбра: (1,2), (1,3), (2,4), (3,5), (5,6), (6,7)
### Дерево (для построения BST)
#Исходный список элементов:  [32, 24, 38, 22, 26, 34, 41]
#Операции:
#- Найти элемент: 26
#- Удалить элемент: 24
### Куча (для Heap Sort)
#Исходный массив:  [32, 24, 38, 22, 26, 34, 41]
## Описание реализации
### Структуры данных
#- Граф: представлен списком смежности (`defaultdict(list)`), а также матрицей смежности (`list[list[int]]`).
#- Бинарное дерево поиска (BST): реализовано в виде класса `Node` с атрибутами `value, left, right`.
#- Куча: реализована на месте (in-place) в массиве для алгоритма Heap Sort.


def heapify(arr, n, i):
    largest = i
    left = 2 * i + 1
    right = 2 * i + 2

    if left < n and arr[left] > arr[largest]:
        largest = left
    if right < n and arr[right] > arr[largest]:
        largest = right

    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)

def heap_sort(arr):
    n = len(arr)

    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)

    for i in range(n - 1, 0, -1):
        arr[i], arr[0] = arr[0], arr[i]
        heapify(arr, i, 0)
```

```python
import networkx as nx
import matplotlib.pyplot as plt

vertices = [(1,2), (1,3), (2,4), (3,5), (5,6), (6,7)]
edges = [(1,2), (1,3), (2,4), (3,5), (5,6), (6,7)]

G_graph = nx.Graph()
G_graph.add_edges_from(edges)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
pos_spring = nx.spring_layout(G_graph, seed=42)
nx.draw(G_graph, pos_spring, with_labels=True, node_color='lightblue', edge_color='gray', node_size=1000, font_weight='bold', font_size=12)
plt.subplot(1, 2, 2)
pos_circular = nx.circular_layout(G_graph)
nx.draw(G_graph, pos_circular, with_labels=True, node_color='lightgreen', edge_color='black', node_size=1000, font_weight='bold', font_size=12)
plt.tight_layout()
plt.savefig('graph_visualization.png', dpi=300)
plt.show()

print(f"Graph nodes: {G_graph.number_of_nodes()}")
print(f"Graph edges: {G_graph.number_of_edges()}")

heap_arr = [32, 24, 38, 22, 26, 34, 41]
n_heap = len(heap_arr)

G_tree = nx.Graph()
for i in range(n_heap):
    left = 2 * i + 1
    right = 2 * i + 2
    if left < n_heap:
        G_tree.add_edge(heap_arr[i], heap_arr[left])
    if right < n_heap:
        G_tree.add_edge(heap_arr[i], heap_arr[right])

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
pos_spring_tree = nx.spring_layout(G_tree, seed=42)
nx.draw(G_tree, pos_spring_tree, with_labels=True, node_color='plum', edge_color='darkred', node_size=1000, font_weight='bold', font_size=12)
plt.subplot(1, 2, 2)
pos_circular_tree = nx.circular_layout(G_tree)
nx.draw(G_tree, pos_circular_tree, with_labels=True, node_color='orange', edge_color='purple', node_size=1000, font_weight='bold', font_size=12)
plt.tight_layout()
plt.savefig('heap_tree_visualization.png', dpi=300)
plt.show()

print(f"Tree nodes: {G_tree.number_of_nodes()}")
print(f"Tree edges: {G_tree.number_of_edges()}")
```
